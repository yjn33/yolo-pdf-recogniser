```python
import os
import random
import shutil


# 훈련 train 디렉토리 경로 설정
img_train_dir = "/home/zen35/Desktop/genimage/data/images/train"
lbl_train_dir = "/home/zen35/Desktop/genimage/data/labels/train"

# 검증(val)용 디렉토리
img_val_dir = "/home/zen35/Desktop/genimage/data/images/val"
lbl_val_dir = "/home/zen35/Desktop/genimage/data/labels/val"



# 디렉토리 생성(없을경우)
os.makedirs(img_val_dir, exist_ok=True)
os.makedirs(lbl_val_dir, exist_ok=True)



# 전체 이미지 리스트
all_images = [f for f in os.listdir(img_train_dir) if f.endswith(('.jpg', '.jpeg', '.png'))]
random.shuffle(all_images)

# 검증 데이터 비율 설정 0.2는 백분율로 20%의미한다
val_ratio = 0.2
val_count = int(len(all_images) * val_ratio)
val_images = all_images[:val_count]



# 이미지 및 라벨 이동
for img_name in val_images:
    # 이미지 이동
    src_img_path = os.path.join(img_train_dir, img_name)
    dst_img_path = os.path.join(img_val_dir, img_name)
    shutil.move(src_img_path, dst_img_path)

    # 라벨 이동
    label_name = os.path.splitext(img_name)[0] + ".txt"
    src_lbl_path = os.path.join(lbl_train_dir, label_name)
    dst_lbl_path = os.path.join(lbl_val_dir, label_name)

    if os.path.exists(src_lbl_path):  # 라벨 파일이 존재할 경우에만 이동
        shutil.move(src_lbl_path, dst_lbl_path)

print(f"{val_count}개의 이미지와 라벨이 val 디렉토리로 이동되었습니다.")
```
