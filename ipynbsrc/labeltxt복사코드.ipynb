```python
import shutil
import os

# 복사할 labels.txt
source_file = "/home/zen35/Documents/1222/1222/labels/3_0.txt"

# 복사할 대상 파일 이름 목록
copy_targets = [f"3_{i}.txt" for i in range(0, 200)]  # → 1_0.txt ~ 1_199.txt

# 복사된 파일 저장 폴더
target_dir = "/home/zen35/Documents/1222/1222/labels/gen3"
os.makedirs(target_dir, exist_ok=True)  # 폴더 없으면 생성

# 복사 시작
if not os.path.exists(source_file):
    print(f"Source file not found: {source_file}")
else:
    for target_file in copy_targets:
        target_path = os.path.join(target_dir, target_file)
        shutil.copy(source_file, target_path)
        print(f"Copied: {source_file} → {target_path}")
```
